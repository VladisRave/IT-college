### **Задача 5: Словари**
Создайте программу для учета товаров в магазине:
```python
products = {
    "яблоки": {"цена": 50, "количество": 100},
    "бананы": {"цена": 70, "количество": 50},
    "апельсины": {"цена": 80, "количество": 75}
}
```
